# Vector Autoregression (VAR)

The Vector Auto Regressive Process is a generalization of the single variable Autoregressive (AR) process to multiple variables. The $\text{VAR}(n)$ </br> 
is defined by,

$
\begin{align}
X_t = \mathcal{M} + \Phi_1 X_{t-1} \Phi_2 X_{t-2} + \cdots + \Phi_n X_{t-n} + \mathcal{E}_t
\end{align}
$

where $t = 1, 2, 3, \ldots, T$, $X_t$ is column vector of length $m$ representing the time series, $\mathcal{M}$ is a $m$ element constant offset column vector, </br>
$\Phi_t$ are $m \times m$ matrices denoting constant coefficients and $\mathcal{E}_t$ an a random column normal vector with zero mean and covariance matrix $\Omega$,

$
\begin{align}
\mathcal{E}_t \sim \text{Normal}(0, \Omega)
\end{align}
$

The $\mathcal{E}_t$ are assumed uncorrelated,

$
\begin{align}
\text{Cov}(\mathcal{E}_t, \mathcal{E}_s) = \left\{
    \begin{array}{ll}
     \Omega  & \mbox{if} t = s \\
    0& \mbox{if} t \ne s
    \end{array}
\right.
\end{align}
$

where,

$
\begin{align}
\Omega = \text{E}(\mathcal{E}_t \mathcal{E}_t^\prime)
\end{align}
$

with $\mathcal{E}_t^\prime$ denoting the transpose of $\mathcal{E}_t$. Explicitly the matrices are given by,

$
\begin{align}
X_t =
\begin{bmatrix}
x^1_t \\
x^2_t \\
\vdots \\
x^m_t \\
\end{bmatrix}
\end{align}
$

$
\begin{align}
\mathcal{M} =
\begin{bmatrix}
\mu_1 \\
\mu_2 \\
\vdots \\
\mu_m \\
\end{bmatrix}
\end{align}
$

$
\begin{align}
\mathcal{E} =
\begin{bmatrix}
\varepsilon^1_t \\
\varepsilon^2_t \\
\vdots \\
\varepsilon^m_t \\
\end{bmatrix}
\end{align}
$

$
\begin{align}
\Phi_i =
\begin{bmatrix}
\phi^i_{11} & \phi^i_{12} & \cdots & \phi^i_{1m} \\
\phi^i_{12} & \phi^i_{22} & \cdots & \phi^i_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^i_{1m} & \phi^i_{1m} & \cdots & \phi^i_{mm} \\
\end{bmatrix}
\end{align}
$

## Examples

$\text{VAR}(1)$ is given by, 

$
\begin{align}
X_t = \mathcal{M} + \Phi_1 X_{t-1} + \mathcal{E}_t
\end{align}
$

where $X_t$, $\mathcal{M}$ and $\mathcal{E}$ are given by the column vector of length $m$ from the previous section and,

$
\begin{align}
\Phi_1 =
\begin{bmatrix}
\phi^1_{11} & \phi^1_{12} & \cdots & \phi^1_{1m} \\
\phi^1_{12} & \phi^1_{22} & \cdots & \phi^1_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^1_{1m} & \phi^1_{1m} & \cdots & \phi^1_{mm} \\
\end{bmatrix}
\end{align}
$

$\text{VAR}(2)$ is given by,

$
\begin{align}
X_t = \mathcal{M} + \Phi_1 X_{t-1} + \Phi_2 X_{t-2} + \mathcal{E}_t
\end{align}
$

where,

$
\begin{align}
\Phi_1 =
\begin{bmatrix}
\phi^1_{11} & \phi^1_{12} & \cdots & \phi^1_{1m} \\
\phi^1_{12} & \phi^1_{22} & \cdots & \phi^1_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^1_{1m} & \phi^1_{1m} & \cdots & \phi^1_{mm} \\
\end{bmatrix}
\end{align}
$

$
\begin{align}
\Phi_2 =
\begin{bmatrix}
\phi^2_{11} & \phi^2_{12} & \cdots & \phi^2_{1m} \\
\phi^2_{12} & \phi^2_{22} & \cdots & \phi^2_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^2_{1m} & \phi^2_{1m} & \cdots & \phi^2_{mm} \\
\end{bmatrix}
\end{align}
$

Finally $\text{VAR}(3)$ is given by,

$
\begin{align}
X_t = \mathcal{M} + \Phi_1 X_{t-1} + \Phi_2 X_{t-2} + \Phi_3 X_{t-3} + \mathcal{E}_t
\end{align}
$

where,

$
\begin{align}
\Phi_1 =
\begin{bmatrix}
\phi^1_{11} & \phi^1_{12} & \cdots & \phi^1_{1m} \\
\phi^1_{12} & \phi^1_{22} & \cdots & \phi^1_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^1_{1m} & \phi^1_{1m} & \cdots & \phi^1_{mm} \\
\end{bmatrix}
\end{align}
$

$
\begin{align}
\Phi_2 =
\begin{bmatrix}
\phi^2_{11} & \phi^2_{12} & \cdots & \phi^2_{1m} \\
\phi^2_{12} & \phi^2_{22} & \cdots & \phi^2_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^2_{1m} & \phi^2_{1m} & \cdots & \phi^2_{mm} \\
\end{bmatrix}
\end{align}
$

$
\begin{align}
\Phi_3 =
\begin{bmatrix}
\phi^3_{11} & \phi^3_{12} & \cdots & \phi^3_{1m} \\
\phi^3_{12} & \phi^3_{22} & \cdots & \phi^3_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^3_{1m} & \phi^3_{1m} & \cdots & \phi^3_{mm} \\
\end{bmatrix}
\end{align}
$

## Companion Form

Consider $\text{VAR}(2)$ from the previous section. To obtain the companion form rewrite the model by stacking the $X_t$ column vector,

$
\begin{align}
X_t =
\begin{bmatrix}
x^1_t \\
x^2_t \\
\vdots \\
x^m_t \\
x^1_{t-1} \\
x^2_{t-1} \\
\vdots \\
x^m_{t-1} \\
\end{bmatrix}
\end{align}
$

and pad the $\mathcal{M}$ and $\mathcal{E}$ zeros in the following manner,

$
\begin{align}
\mathcal{M} =
\begin{bmatrix}
\mu_1 \\
\mu_2 \\
\vdots \\
\mu_m \\
0 \\
0 \\
\vdots \\
0
\end{bmatrix}
\end{align}
$

$
\begin{align}
\mathcal{E} =
\begin{bmatrix}
\varepsilon^1_t \\
\varepsilon^2_t \\
\vdots \\
\varepsilon^m_t \\
0 \\
0 \\
\vdots \\
0
\end{bmatrix}
\end{align}
$

finally rewrite the coefficient matrix as follows,

$
\begin{align}
\Phi =
\begin{bmatrix}
\phi^1_{11} & \phi^1_{12} & \cdots & \phi^1_{1m} & \phi^2_{11} & \phi^2_{12} & \cdots & \phi^2_{1m} \\
\phi^1_{12} & \phi^1_{22} & \cdots & \phi^1_{2m} & \phi^2_{12} & \phi^2_{22} & \cdots & \phi^2_{2m}\\
\vdots & \vdots & \ddots & \vdots & \vdots & \vdots & \ddots & \vdots \\
\phi^1_{1m} & \phi^1_{1m} & \cdots & \phi^1_{mm} & \phi^2_{1m} & \phi^2_{1m} & \cdots & \phi^2_{mm}\\
1 & 0 & \cdots & 0 & 0 & 0 & \cdots & 0\\
0 & 1 & \cdots & 0 & 0 & 0 & \cdots & 0\\
\vdots & \vdots & \ddots & \vdots & \vdots & \vdots & \ddots & \vdots\\
0 & 0 & \cdots & 1 & 0 & 0 & \cdots & 0\\
\end{bmatrix}
= \begin{bmatrix}
\Phi_1 & \Phi_2 \\
\mathbb{1} & 0
\end{bmatrix}
\end{align}
$

Using this representation $\text{VAR}(2)$ can be rewritten as,

$
\begin{align}
X_t = \mathcal{M} + \Phi X_{t-1} + \mathcal{E}_t
\end{align}
$

which has the form of $\text{VAR}(1)$. This is called the companion form of $\text{VAR}(2)$. It can be shown that an arbitrary order $\text{VAR}(n)$ </br>
model can be written companion form.
