# Vector Autoregression (VAR)

The Vector Auto Regressive Process is a generalization of the single variable Autoregressive (AR) process to multiple variables. The $\text{VAR}(n)$ </br> 
is defined by,

$
\begin{align}
X_t = \mathcal{M} + \Phi_1 X_{t-1} \Phi_2 X_{t-2} + \cdots + \Phi_n X_{t-n} + \mathcal{E}_t
\end{align}
\tag{1}
$

where $t = 1, 2, 3, \ldots, T$, $X_t$ is column vector of length $m$ representing the time series, $\mathcal{M}$ is a $m$ element constant offset column vector, $\Phi_t$ are $m \times m$ matrices denoting constant coefficients and $\mathcal{E}_t$ an a random column normal vector with zero mean and covariance matrix $\Omega$,

$
\begin{align}
\mathcal{E}_t \sim \text{Normal}(0, \Omega)
\end{align}
\tag{2}
$

The $\mathcal{E}_t$ are assumed uncorrelated, i.e. if $t=s$, 

$
\begin{align}
\text{Cov}(\mathcal{E}_t, \mathcal{E}_s) = \Omega 
\end{align}
\tag{3}
$

and $0$ if $t \ne s$, where,

$
\begin{align}
\Omega = \text{E}(\mathcal{E}_t \mathcal{E}_t^\prime)
\end{align}
\tag{4}
$

with $\mathcal{E}_t^\prime$ denoting the transpose of $\mathcal{E}_t$. Explicitly the matrices are given by,

$
\begin{align}
X_t =
\begin{bmatrix}
x^1_t \\
x^2_t \\
\vdots \\
x^m_t \\
\end{bmatrix}
\end{align}
\tag{5}
$

$
\begin{align}
\mathcal{M} =
\begin{bmatrix}
\mu_1 \\
\mu_2 \\
\vdots \\
\mu_m \\
\end{bmatrix}
\end{align}
\tag{6}
$

$
\begin{align}
\mathcal{E}_t =
\begin{bmatrix}
\varepsilon^1_t \\
\varepsilon^2_t \\
\vdots \\
\varepsilon^m_t \\
\end{bmatrix}
\end{align}
\tag{7}
$

$
\begin{align}
\Phi_i =
\begin{bmatrix}
\phi^i_{11} & \phi^i_{12} & \cdots & \phi^i_{1m} \\
\phi^i_{12} & \phi^i_{22} & \cdots & \phi^i_{2m} \\
\vdots & \vdots & \ddots & \vdots \\
\phi^i_{m1} & \phi^i_{m2} & \cdots & \phi^i_{mm} \\
\end{bmatrix}
\end{align}
\tag{8}
$

## Simulation of VAR Process

The Cholesky method is used to simulate correlated Gaussian processes, $\mathbb{Z}_n$ as a liner transformation of independent unit normal random variables. </br>
Let $\mathbb{B}_n$ denote a column vector of $n$ independent unit normal random variables,

$
\begin{align}
\mathbb{B}_n =
\begin{bmatrix}
B_1 \\
B_2 \\
\vdots \\
B_n \\
\end{bmatrix}
\end{align}
\tag{18}
$

where $B_i \sim \text{Normal}(0, 1)$. Let $L_n$ denote an $n\times n$ matrix where,

$
\begin{align}
\mathbb{Z}_n = L_n \mathbb{B}_n
\end{align}
\tag{19}
$

Let $\Omega_n$ denote the covariance of the random vector $\mathbb{Z}_n$ and $\mathbb{Z}_n^\prime$ its transpose. It follows that,</br>

$
\begin{align}
\text{E}\left(\mathbb{Z}_n, \mathbb{Z}_n^\prime \right) = \Omega_n
\end{align}
\tag{20}
$

Since the elements of $\mathbb{B}_n$ are unit normal random variables,

$
\begin{align}
\text{E}\left( \mathbb{B}_n \mathbb{B}_n^\prime\right) = \mathbb{1}_n
\end{align}
\tag{21}
$

where $\mathbb{1}_n $ is an $n\times n$ identity matrix. From the definition of the covariance matrix it is easy to show that it is symmetric, $\Omega = \Omega^\prime$, and that</br>
it is positive definite $X_n^\prime \Omega_n^{-1}X_n \ge 0$ for any real vector $X_n$. It follows that, 

$
\begin{align}
\Omega &= \text{E}\left(\mathbb{Z}_n, \mathbb{Z}_n^\prime \right) \\
&= \text{E}\left[\left(L_n \mathbb{B}_n\right)  \left(L_n \mathbb{B}_n \right)^\prime  \right] \\
& = \text{E}\left( L_n \mathbb{B}_n \mathbb{B}_n^\prime L^\prime \right)
\end{align}
\tag{22}
$

recall that $\text{E}\left( \mathbb{B}_b \mathbb{B}_n^\prime \right) = \mathbb{1}_n$ so it follows that,

$
\begin{align}
\Omega_n &= L_n\text{E}\left( \mathbb{B}_n \mathbb{B}_n^\prime \right) L_n^\prime \\
&= L_n L_n^\prime
\end{align}
\tag{23}
$

Recall Cholesky decomposition is a method for factorization of a symmetric positive definite matrix into the product of a lower diagonal matrix and its transpose. </br>
Since $\Omega_n =L_n L_n^\prime$ Cholesky decomposition can be used to compute $L_n$ by assuming it is a lower triangular matrix. It follows that the simulated Gaussian process </br>
$\mathbb{Z}_n$ can be computed from,

$
\begin{align}
\mathbb{Z}_n = L_n\mathbb{B}_n
\end{align}
\tag{24}
$

where $\mathbb{B}_n \sim \text{Normal}(0,n)$ and $\mathbb{Z}_n \sim \text{Normal}(0, \Omega)$. Thus the procedure for simulation of a correlated Gaussian process is to assume a </br>
correlation matrix $\Omega$ then compute $L_n$ using Cholesky decomposition then the simulated vector is given by, $\mathbb{Z}_n = L_n \mathbb{B}_n$. </br>
It has been assumed that the simulated process has zero mean. A mean value can be incorporated in the following matter. If the mean column vector is given by,

$
\begin{align}
\mathcal{M}_n =
\begin{bmatrix}
\mu_1 \\
\mu_2 \\
\vdots \\
\mu_n \\
\end{bmatrix}
\end{align}
\tag{25}
$

Then,

$
\begin{align}
\mathbb{Z}_n - \mathcal{M}_n = L_n \mathbb{B}_n
\end{align}
\tag{26}
$

After generating the random component using the Cholesky method a step in $\text{VAR}(n)$ process is computed using,

$
\begin{align}
X_t = \mathcal{M} + \Phi_1 X_{t-1} + \Phi_2 X_{t-2} + \cdots + \Phi_n X_{t-n} + \mathcal{E}_t
\end{align}
\tag{27}
$

## Helpers

```python
def comparison_plot(xt, title, ylim=(-2.5, 2.5), xlim=(0, 1000)):
    ylabels = [f"$S_{i+1}(t)$" for i in range(len(xt))]
    stack(xt, ylim=ylim, xlim=xlim, title=title, ylabels=ylabels, xlabel=r"$t$", figsize=(10, 6), title_offset=0.2)

def cumu_mean_plot(t, xt, mean, title):
    _, cumu_mean = stats.compute_cumu_mean(t, xt)
    stationary_mean = numpy.full(len(t), mean)
    fcurve(data=cumu_mean, func=stationary_mean, x=t, fx=t, title_offset=0.025, plot_axis_type=PlotType.XLOG, xlabel=r"$t$", ylabel=r"$\mu_t$",
           labels=["Simulation", r"$\lim_{t \to \infty}\, \mu_t$"], title=title)

def cumu_var_plot(t, xt, sd, title):
    _, cumu_sd = stats.compute_cumu_var(t, xt)
    limit_sd = numpy.full(len(t), sd)
    fcurve(data=cumu_sd, func=limit_sd, x=t, fx=t, title_offset=0.025, plot_axis_type=PlotType.XLOG, xlabel=r"$t$", ylabel=r"$\sigma_t^2$",
           labels=["Simulation", r"$\lim_{t \to \infty}\, \sigma^2_t$"], title=title)
    
def cumu_cov_plot(t, xt, yt, cov, title):
    _, cumu_cov = stats.compute_cumu_cov(t,xt, yt)
    limit_cov = numpy.full(len(t), cov)
    fcurve(data=cumu_cov, func=limit_cov, x=t, fx=t, title_offset=0.025, plot_axis_type=PlotType.XLOG, xlabel=r"$t$", ylabel=r"$\gamma_t$",
           labels=["Simulation", r"$\lim_{t \to \infty}\, \gamma_t$"], title=title)

def acf_plot(t, xt, var_acf, nlags, title):
    lag, acf = stats.compute_acf(t, xt, nlags=nlags)
    fpoints(data=acf, func=var_acf, x=lag, fx=lag, title_offset=0.025, xlabel=r"$\tau$", ylabel=r"$\rho_\tau$",
           labels=["Simulation", r"$\lim_{t \to \infty}\, \rho_{t, \tau}$"], title=title)
```

## $\text{VAR}(1)$ Simulations

### Uncoupled $\text{AR}(1)$

```python
phi = numpy.array([
        numpy.matrix([[0.25, 0.0],
                      [0.0, 0.25]])])
mu = numpy.array([0.0, 0.0])
omega = numpy.matrix([[1.0, 0.0],
                      [0.0, 1.0]])

npts=5000
nlag=11
var.compute_is_stationary(phi)
```

```python
t, xt = var.create_source(phi, Ω=omega, μ=mu, npts=npts)
```

```python
title = f"VAR(1) Simulation of Independent Stationary AR(1) Processes"
comparison_plot(xt, title, ylim=(-3.5, 3.5))
```

```python
mean = var.compute_mean(phi, mu)
```

```python
title = f"$S_1(t)$ Convergence of Mean to $t\\rightarrow\infty$ Limit: {mean[0,0]:.2f}"
cumu_mean_plot(t, xt[0], mean[0,0], title)
```

```python
title = f"$S_2(t)$ Convergence of Mean to $t\\rightarrow\infty$ Limit: {mean[1,0]:.2f}"
cumu_mean_plot(t, xt[1], mean[1,0], title)
```

```python
cov = var.compute_cov(phi, omega)
cov
```

```python
title = f"$S_1(t)$ Convergence of Variance to $t\\rightarrow\infty$ Limit: {cov[0,0]:.2f}"
cumu_var_plot(t, xt[0], cov[0,0], title)
```

```python
title = f"$S_2(t)$ Convergence of Variance to $t\\rightarrow\infty$ Limit: {cov[1,1]:.2f}"
cumu_var_plot(t, xt[1], cov[1,1], title)
```

```python
title = f"Convergence of $S_2(t)$ $S_1(t)$ Covariance to $t\\rightarrow\infty$ Limit: {cov[0,1]:.2f}"
cumu_cov_plot(t, xt[0], xt[1], cov[0,1], title)
```

```python
tau, acov = var.compute_acov(phi, omega, nlag=nlag)
```

```python
acf = numpy.array([acov[i,0,0] for i in range(len(acov))]) / acov[0,0,0]
title = f"Convergence of $S_1(t)$ ACF to $t\\rightarrow\infty$ Limit"
acf_plot(t, xt[0], acf, nlag, title)
```

```python
acf = numpy.array([acov[i,1,1] for i in range(len(acov))]) / acov[0,1,1]
title = f"Convergence of $S_2(t)$ ACF to $t\\rightarrow\infty$ Limit"
acf_plot(t, xt[1], acf, nlag, title)
```
