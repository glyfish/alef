# Vector Error Correction Model (VECM) Parameter Estimation

Recall that the VECM is given by,

$
\begin{align}
\Delta X_t &= A_1  X_{t-1} + A_2 X_{t-1}+ \cdots + A_n \Delta X_{t - n + 1}  + \mathcal{E}_t \\
           &= A_1 X_{t-1} + \sum_{i=2}^n A_i \Delta X_{t - i + 1}  + \mathcal{E}_t
\end{align}
\tag{1}
$

where,

$
\begin{align}
\Delta X_t = \left[\Delta x_{1,t}, \Delta x_{2,t}, \ldots, \Delta x_{k,t}\right]^\prime
\end{align}
\tag{2}
$

and $A_i$ are square $k\times k$ matrices and $\mathcal{E}_t$ is a serially uncorrelated random column vector of length $k$ with distribution

$
\begin{align}
\mathcal{E}_t \sim \text{Normal}(0,\Omega)
\end{align}
\tag{3}
$ 

and covariance matrix for $t=s$,

$
\begin{align}
\text{Cov}\left(\mathcal{E}_t \mathcal{E}_s\right) = \Omega
\end{align}
\tag{4}
$ 

and for $t\ne s$

$
\begin{align}
\text{Cov}\left(\mathcal{E}_t \mathcal{E}_s\right) = 0
\end{align}
\tag{5}
$ 

Let $A_1$ = \alpha\beta^\prime$ then cointegration requires

$
\begin{align}
\text{rank}\left(\alpha \beta^\prime\right) = r < k
\end{align}
\tag{6}
$ 

here $\alpha$ has order $k\times r$ and $\beta^\prime$ has order $r\times k$

## Canonical Correlation Analysis

Canonical Correlation Analysis is used to estimate the VECM parameters. Here an overview will be given. Consider two random vectors where one has length $p$ and the other has length $q$,

$
\begin{align}
X &= \left[x_1, x_2, x_3, \ldots, x_p\right]^\prime \\
Y &= \left[y_1, y_2, y_3, \ldots, y_q\right]^\prime
\end{align}
\tag{7}
$

let,

$
\begin{align}
Z = \begin{bmatrix}
X \\
Y
\end{bmatrix}
\end{align}
\tag{8}
$

Assume that $\text{E}\left[Z\right]=0$ then,

$
\begin{align}
\Sigma &= \text{Var}\left[Z\right] \\
&= \text{E}\left[Z Z^\prime\right] \\
&= \text{E}
   \left[ 
   \begin{bmatrix}
   X \\
   Y
   \end{bmatrix}
   \begin{bmatrix}
   X^\prime & Y^\prime
   \end{bmatrix}
   \right] \\
&= \begin{bmatrix}
    \text{E}\left[XX^\prime\right] & \text{E}\left[XY^\prime\right] \\
    \text{E}\left[YX^\prime\right] & \text{E}\left[YY^\prime\right]
   \end{bmatrix}
\end{align}
\tag{9}
$

Further let,

$
\begin{align}
\Sigma_{xx} &= \text{E}\left[XX^\prime\right] \\
\Sigma_{xy} &= \text{E}\left[XY^\prime\right] \\
\Sigma_{yx} &= \text{E}\left[YX^\prime\right] \\
\Sigma_{yy} &= \text{E}\left[YY^\prime\right] \\
\end{align}
\tag{10}
$

Now $\Sigma_{xx}$ has order $p\times p$, $\Sigma_{yy}$ has order $q\times q$, $\Sigma_{xy}$ has order $p\times q$ and $\Sigma_{yx}$ has order $q\times p$. It follows that \Sigma is a square matrix with order $\left(q + p\right)\times\left(q + p\right)$. Finally

$
\begin{align}
\Sigma =
\begin{bmatrix}
\Sigma_{xx} & \Sigma{xy} \\
\Sigma_{yy} & \Sigma{yx}
\end{bmatrix}
\end{align}
\tag{11}
$

This is called the partition of the covariance matrix. Consider the transpose,

$
\begin{align}
\Sigma^\prime &= \left(\text{E}\left[ZZ^\prime\right]\right)^\prime \\
&= \text{E}\left[\left(ZZ^\prime\right)^\prime\right] \\
&= \text{E}\left[ZZ^\prime\right] \\
& = \Sigma
\end{align}
\tag{12}
$

which is a well known property of covariance matrices. Similarly it can be shown that,

$
\begin{align}
\Sigma_{xx}^\prime &= \Sigma_{xx} & \Sigma_{yy}^\prime &= \Sigma_{yy} \\
\Sigma_{xy}^\prime &= \Sigma_{yx} & \Sigma_{yx}^\prime &= \Sigma_{xy}
\end{align}
\tag{13}
$

Consider the linear combinations,

$
\begin{align}
U &= A^\prime X \\
V &= B^\prime Y \\
\end{align}
\tag{14}
$

where $A$ and $B$ are constant vectors. It follows that

$
\begin{align}
\text{Var}\left(U\right) &= A^\prime \Sigma_{xx} A \\
\text{Var}\left(V\right) &= B^\prime \Sigma_{yy} B
\end{align}
\tag{15}
$

and,

$
\begin{align}
\text{Cov}\left(UV\right) &= A^\prime \Sigma_{xy} B \\
\text{Cov}\left(VU\right) &= B^\prime \Sigma_{yx} A
\end{align}
\tag{15}
$

Combining these expressions gives the correlation coefficients,

$
\begin{align}
\rho_{uv} &= \frac{\text{Cov}\left(UV\right)}{\sqrt{\text{Var}\left(U\right)\text{Var}\left(V\right)}} \\
          &= \frac{A^\prime \Sigma_{xy} B}{\sqrt{\left(A^\prime \Sigma_{xx} A\right)\left(B^\prime \Sigma_{yy} B\right)}} \\
\rho_{vu} &= \frac{\text{Cov}\left(VU\right)}{\sqrt{\text{Var}\left(U\right)\text{Var}\left(V\right)}} \\
          &= \frac{B^\prime \Sigma_{yx} A}{\sqrt{\left(A^\prime \Sigma_{xx} A\right)\left(B^\prime \Sigma_{yy} B\right)}}
\end{align}
\tag{16}
$

The canonical correlation is defined as the vectors $A$ and $B$ that maximize $\rho_{UV}$ subject to the constraints,

$
\begin{align}
A^\prime \Sigma_{xx} A &= 1 \\
B^\prime \Sigma_{yy} B &= 1
\end{align}
\tag{17}
$

First consider $\rho_{UV}$, Lagrange multipliers cab be used to maximize $A^\prime \Sigma_{xy} B$ subject to the constraints specified by Equation $(17)$. The Lagrangian is given by,

$
\begin{align}
L = A^\prime \Sigma_{xy} B - \alpha \left(A^\prime \Sigma_{xx} A - 1\right) - \beta\left(B^\prime \Sigma_{yy} B - 1\right)
\end{align}
\tag{18}
$

where $\alpha$ and $\beta$ are Lagrange multipliers. After much algebra it can be shown that $\alpha=\beta$ and that $\alpha$ can be interpreted as the maximum value of the correlation,

$
\begin{align}
A^\prime \Sigma_{xy} B =  B^\prime \Sigma_{yx} A = \alpha
\end{align}
\tag{19}
$

and that $A$ and $B$ are the eigenvectors corresponding to the largest eigenvalue that are solutions to the equations,

$
\begin{align}
\left(\Sigma^{-1}_{xx}\Sigma_{xy}\Sigma^{-1}_{yy}\Sigma_{yx} - \alpha^2 \mathbb{I}\right)A &= 0\\
\left(\Sigma^{-1}_{yy}\Sigma_{yx}\Sigma^{-1}_{xx}\Sigma_{xy} - \alpha^2 \mathbb{I}\right)B &= 0
\end{align}
\tag{20}
$

Let $A_1$ and $B_1$ denote the eigenvectors corresponding to the largest eigenvalue $\alpha_1^2$ so the the first canonical correlation pair is given by,

$
\begin{align}
U_1 &= A_1 X \\
V_1 &= B_1 Y
\end{align}
\tag{21}
$

with correlation coefficient $\alpha_1$. The second pair of canonical correlates is obtained from the second largest eigenvalue $\alpha_2$ and its eigenvectors $A_2$ and $B_2$ by maximizing correlation,

$
\begin{align}
\rho_2 = A^\prime_2 \Sigma_{xy} B_2
\end{align}
\tag{22}
$

subject to the constraints,

$
\begin{align}
A_2^\prime \Sigma_{xx} A_2 &= 1 \\
B_2^\prime \Sigma_{yy} B_2 &= 1
\end{align}
\tag{23}
$

and additionally the second canonical pair is uncorrelated with the first so the following constraints are also applied,

$
\begin{align}
A_2^\prime \Sigma_{xx} A_1 &= 0 \\
B_2^\prime \Sigma_{yy} B_1 &= 0
\end{align}
\tag{24}
$

After much algebra it is seen that the second canonical pair satisfies the same equation as the first but for the second largest eigenvalue and it eigenvectors.

$
\begin{align}
\left(\Sigma^{-1}_{xx}\Sigma_{xy}\Sigma^{-1}_{yy}\Sigma_{yx} - \alpha_2^2 \mathbb{I}\right)A_2 &= 0\\
\left(\Sigma^{-1}_{yy}\Sigma_{yx}\Sigma^{-1}_{xx}\Sigma_{xy} - \alpha_2^2 \mathbb{I}\right)B_2 &= 0
\end{align}
\tag{25}
$

The third canonical pair is constructed from the third largest eigenvalue and its eigenvectors with the constraint that they are uncorrelated to the first and second pairs and so on until the eigenvalues are exhausted.

## Reduced Rank Regression

The VECM can be converted to the Anderson model form used in reduced rank regression by defining the following variables,

$
\begin{align}
\mathbb{Y}_t &= \Delta X_t\\
\mathbb{X}_t &= X_{t-1} \\
\mathbb{Z}_t &= \left[\Delta X^\prime_{t-1}\ \Delta X^\prime_{t-2}\ \cdots \Delta X^\prime_{t-n+1}\right]^\prime \\
\mathbb{D} &= \left[A_2\ A_3\ \cdots\ A_n\right]
\end{align}
\tag{26}
$

where $mathbb{Y}_t$ and $\mathbb{X}_t$ are column vectors of order $k\times 1$. $\mathbb{Z}_t$ is a block column vector of $n-1$ vectors of length $k$ and $\mathbb{D}$ is a block row vector of length $n-1$ of square matrices of order $k\times k$. Using these variables the VECM becomes,

$
\begin{align}
\mathbb{Y}_t &= \alpha \beta^\prime \mathbb{X}_t + \mathbb{D}\mathbb{Z}_t +  \mathcal{E}_t
\end{align}
\tag{27}
$

Further define,

$
\begin{align}
\Sigma_{yx} &= \text{E}\left[\mathbb{Y}_t \mathbb{X}_t^\prime\right] \\
\Sigma_{xy} &= \text{E}\left[\mathbb{X}_t \mathbb{Y}_t^\prime\right] = \Sigma_{yx}^\prime \\
\Sigma_{zy} &= \text{E}\left[\mathbb{Z}_t \mathbb{Y}_t^\prime\right] = \Sigma_{yz}^\prime \\
\Sigma_{xz} &= \text{E}\left[\mathbb{X}_t \mathbb{Z}_t^\prime\right] = \Sigma_{zx}^\prime \\
\end{align}
\tag{28}
$

$
\begin{align}
\Sigma_{xx} &= \text{E}\left[\mathbb{Y}_t \mathbb{X}_t^\prime\right] \\
\Sigma_{yy} &= \text{E}\left[\mathbb{X}_t \mathbb{Y}_t^\prime\right] \\
\Sigma_{zz} &= \text{E}\left[\mathbb{Z}_t \mathbb{Y}_t^\prime\right]
\end{align}
\tag{29}
$

Define the transformations,

$
\begin{align}
\mathbb{X}_t^* &= \mathbb{X}_t - \Sigma_{xz}\Sigma^{-1}_{zz}\mathbb{Z}_t \\
\mathbb{D}^* &= \mathbb{D} + \alpha\beta^\prime\Sigma_{xz}\Sigma^{-1}_{zz}
\end{align}
\tag{30}
$

It is seen that $\mathbb{X}_t^*$ is the residual of the OLS regression of $\mathbb{X}_t$ on $\mathbb{Z}_t$ where $\Sigma_{xz}\Sigma^{-1}$ is the regression coefficient. Note that $\mathbb{X}_t^*$  and $\mathbb{Z}_t$ are uncorrelated,

$
\begin{align}
\text{E}\left[\mathbb{X}_t^* \mathbb{Z}^\prime_t\right] = \Sigma_{xz} - \Sigma_{xz}\Sigma_{zz}^{-1} \Sigma_{zz} = 0
\end{align}
\tag{31}
$

It follows that,

$
\begin{align}
\mathbb{Y}_t &= \alpha \beta^\prime \mathbb{X}^*_t + \mathbb{D}^*\mathbb{Z}_t +  \mathcal{E}_t
\end{align}
\tag{32}
$

further let,

$
\begin{align}
\mathbb{Y}^*_t &= \mathbb{Y}_t - \mathbb{D}^*\mathbb{Z}_t
\end{align}
\tag{33}
$

Regressing $\mathbb{Y}_t$ on $\mathbb{Z}_t$ using OLS gives,

$
\begin{align}
\mathbb{D}^* = \Sigma_{yz}\Sigma_{zz}^{-1}
\end{align}
\tag{31}
$

so $\mathbb{Y}^*_t$ is the residual of the regression of $\mathbb{Y}_t$ on  $\mathbb{Z}_t$. Bringing things together gives,

$
\begin{align}
\mathbb{Y}^*_t &= \alpha\beta^\prime \mathbb{X}^*_t + \mathcal{E}_t \\
\mathbb{D} &= \mathbb{D}^* - \alpha\beta^\prime\Sigma_{xz}\Sigma_{zz}^{-1} \\
&= \Sigma_{yz}\Sigma_{zz} - \alpha\beta^\prime\Sigma_{xz}\Sigma_{zz}^{-1}
\end{align}
\tag{33}
$
