# Vector Error Correction Model (VECM) Parameter Estimation

Recall that the VECM is given by,

$
\begin{align}
\Delta X_t &= A_1  X_{t-1} + A_2 X_{t-1}+ \cdots + A_n \Delta X_{t - n + 1}  + \mathcal{E}_t \\
           &= A_1 X_{t-1} + \sum_{i=2}^n A_i \Delta X_{t - i + 1}  + \mathcal{E}_t
\end{align}
\tag{1}
$

where,

$
\begin{align}
\Delta X_t = \left[\Delta x_{1,t}, \Delta x_{2,t}, \ldots, \Delta x_{k,t}\right]^\prime
\end{align}
\tag{2}
$

and $A_i$ are square $k\times k$ matrices and $\mathcal{E}_t$ is a serially uncorrelated random column vector of length $k$ with distribution

$
\begin{align}
\mathcal{E}_t \sim \text{Normal}(0,\Omega)
\end{align}
\tag{3}
$ 

and covariance matrix for $t=s$,

$
\begin{align}
\text{Cov}\left(\mathcal{E}_t \mathcal{E}_s\right) = \Omega
\end{align}
\tag{4}
$ 

and for $t\ne s$

$
\begin{align}
\text{Cov}\left(\mathcal{E}_t \mathcal{E}_s\right) = 0
\end{align}
\tag{5}
$ 

Let $A_1 = \alpha\beta^\prime$ then cointegration requires

$
\begin{align}
\text{rank}\left(\alpha \beta^\prime\right) = r < k
\end{align}
\tag{6}
$ 

here $\alpha$ has order $k\times r$ and $\beta^\prime$ has order $r\times k$

## Canonical Correlation Analysis

Canonical Correlation Analysis is used to estimate the VECM parameters. Here an overview will be given. Consider two random vectors where one has length $p$ and the other has length $q$,

$
\begin{align}
X &= \left[x_1, x_2, x_3, \ldots, x_p\right]^\prime \\
Y &= \left[y_1, y_2, y_3, \ldots, y_q\right]^\prime
\end{align}
\tag{7}
$

let,

$
\begin{align}
Z = \begin{bmatrix}
X \\
Y
\end{bmatrix}
\end{align}
\tag{8}
$

Assume that $\text{E}\left[Z\right]=0$ then,

$
\begin{align}
\Sigma &= \text{Var}\left[Z\right] \\
&= \text{E}\left[Z Z^\prime\right] \\
&= \text{E}
   \left[ 
   \begin{bmatrix}
   X \\
   Y
   \end{bmatrix}
   \begin{bmatrix}
   X^\prime & Y^\prime
   \end{bmatrix}
   \right] \\
&= \begin{bmatrix}
    \text{E}\left[XX^\prime\right] & \text{E}\left[XY^\prime\right] \\
    \text{E}\left[YX^\prime\right] & \text{E}\left[YY^\prime\right]
   \end{bmatrix}
\end{align}
\tag{9}
$

Further let,

$
\begin{align}
\Sigma_{xx} &= \text{E}\left[XX^\prime\right] \\
\Sigma_{xy} &= \text{E}\left[XY^\prime\right] \\
\Sigma_{yx} &= \text{E}\left[YX^\prime\right] \\
\Sigma_{yy} &= \text{E}\left[YY^\prime\right] \\
\end{align}
\tag{10}
$

Now $\Sigma_{xx}$ has order $p\times p$, $\Sigma_{yy}$ has order $q\times q$, $\Sigma_{xy}$ has order $p\times q$ and $\Sigma_{yx}$ has order $q\times p$. It follows that \Sigma is a square matrix with order $\left(q + p\right)\times\left(q + p\right)$. Finally

$
\begin{align}
\Sigma =
\begin{bmatrix}
\Sigma_{xx} & \Sigma{xy} \\
\Sigma_{yy} & \Sigma{yx}
\end{bmatrix}
\end{align}
\tag{11}
$

This is called the partition of the covariance matrix. Consider the transpose,

$
\begin{align}
\Sigma^\prime &= \left(\text{E}\left[ZZ^\prime\right]\right)^\prime \\
&= \text{E}\left[\left(ZZ^\prime\right)^\prime\right] \\
&= \text{E}\left[ZZ^\prime\right] \\
& = \Sigma
\end{align}
\tag{12}
$

which is a well known property of covariance matrices. Similarly it can be shown that,

$
\begin{align}
\Sigma_{xx}^\prime &= \Sigma_{xx} & \Sigma_{yy}^\prime &= \Sigma_{yy} \\
\Sigma_{xy}^\prime &= \Sigma_{yx} & \Sigma_{yx}^\prime &= \Sigma_{xy}
\end{align}
\tag{13}
$

Consider the linear combinations,

$
\begin{align}
U &= A^\prime X \\
V &= B^\prime Y \\
\end{align}
\tag{14}
$

where $A$ and $B$ are constant vectors. It follows that

$
\begin{align}
\text{Var}\left(U\right) &= A^\prime \Sigma_{xx} A \\
\text{Var}\left(V\right) &= B^\prime \Sigma_{yy} B
\end{align}
\tag{15}
$

and,

$
\begin{align}
\text{Cov}\left(UV\right) &= A^\prime \Sigma_{xy} B \\
\text{Cov}\left(VU\right) &= B^\prime \Sigma_{yx} A
\end{align}
\tag{15}
$

Combining these expressions gives the correlation coefficients,

$
\begin{align}
\rho_{uv} &= \frac{\text{Cov}\left(UV\right)}{\sqrt{\text{Var}\left(U\right)\text{Var}\left(V\right)}} \\
          &= \frac{A^\prime \Sigma_{xy} B}{\sqrt{\left(A^\prime \Sigma_{xx} A\right)\left(B^\prime \Sigma_{yy} B\right)}} \\
\rho_{vu} &= \frac{\text{Cov}\left(VU\right)}{\sqrt{\text{Var}\left(U\right)\text{Var}\left(V\right)}} \\
          &= \frac{B^\prime \Sigma_{yx} A}{\sqrt{\left(A^\prime \Sigma_{xx} A\right)\left(B^\prime \Sigma_{yy} B\right)}}
\end{align}
\tag{16}
$

The canonical correlation is defined as the vectors $A$ and $B$ that maximize $\rho_{UV}$ subject to the constraints,

$
\begin{align}
A^\prime \Sigma_{xx} A &= 1 \\
B^\prime \Sigma_{yy} B &= 1
\end{align}
\tag{17}
$

First consider $\rho_{uv}$, Lagrange multipliers cab be used to maximize $A^\prime \Sigma_{xy} B$ subject to the constraints specified by Equation $(17)$. The Lagrangian is given by,

$
\begin{align}
L = A^\prime \Sigma_{xy} B - \alpha \left(A^\prime \Sigma_{xx} A - 1\right) - \beta\left(B^\prime \Sigma_{yy} B - 1\right)
\end{align}
\tag{18}
$

where $\alpha$ and $\beta$ are Lagrange multipliers. After much algebra it can be shown that $\alpha=\beta$ and that $\alpha$ can be interpreted as the maximum value of the correlation,

$
\begin{align}
A^\prime \Sigma_{xy} B =  B^\prime \Sigma_{yx} A = \alpha
\end{align}
\tag{19}
$

and that $A$ and $B$ are the eigenvectors corresponding to the largest eigenvalue that are solutions to the equations,

$
\begin{align}
\left(\Sigma^{-1}_{xx}\Sigma_{xy}\Sigma^{-1}_{yy}\Sigma_{yx} - \alpha^2 \mathbb{I}\right)A &= 0\\
\left(\Sigma^{-1}_{yy}\Sigma_{yx}\Sigma^{-1}_{xx}\Sigma_{xy} - \alpha^2 \mathbb{I}\right)B &= 0
\end{align}
\tag{20}
$

Let $A_1$ and $B_1$ denote the eigenvectors corresponding to the largest eigenvalue $\alpha_1^2$ so the the first canonical correlation pair is given by,

$
\begin{align}
U_1 &= A_1 X \\
V_1 &= B_1 Y
\end{align}
\tag{21}
$

with correlation coefficient $\alpha_1$. The second pair of canonical correlates is obtained from the second largest eigenvalue $\alpha_2$ and its eigenvectors $A_2$ and $B_2$ by maximizing correlation,

$
\begin{align}
\rho_2 = A^\prime_2 \Sigma_{xy} B_2
\end{align}
\tag{22}
$

subject to the constraints,

$
\begin{align}
A_2^\prime \Sigma_{xx} A_2 &= 1 \\
B_2^\prime \Sigma_{yy} B_2 &= 1
\end{align}
\tag{23}
$

and additionally the second canonical pair is uncorrelated with the first so the following constraints are also applied,

$
\begin{align}
A_2^\prime \Sigma_{xx} A_1 &= 0 \\
B_2^\prime \Sigma_{yy} B_1 &= 0
\end{align}
\tag{24}
$

After much algebra it is seen that the second canonical pair satisfies the same equation as the first but for the second largest eigenvalue and it eigenvectors.

$
\begin{align}
\left(\Sigma^{-1}_{xx}\Sigma_{xy}\Sigma^{-1}_{yy}\Sigma_{yx} - \alpha_2^2 \mathbb{I}\right)A_2 &= 0\\
\left(\Sigma^{-1}_{yy}\Sigma_{yx}\Sigma^{-1}_{xx}\Sigma_{xy} - \alpha_2^2 \mathbb{I}\right)B_2 &= 0
\end{align}
\tag{25}
$

The third canonical pair is constructed from the third largest eigenvalue and its eigenvectors with the constraint that they are uncorrelated to the first and second pairs and so on until the eigenvalues are exhausted.

## Reduced Rank Regression

The VECM can be converted to the Anderson model form used in reduced rank regression by defining the following variables,

$
\begin{align}
\mathbb{Y}_t &= \Delta X_t\\
\mathbb{X}_t &= X_{t-1} \\
\mathbb{Z}_t &= \left[\Delta X^\prime_{t-1}\ \Delta X^\prime_{t-2}\ \cdots \Delta X^\prime_{t-n+1}\right]^\prime \\
\mathbb{D} &= \left[A_2\ A_3\ \cdots\ A_n\right]
\end{align}
\tag{26}
$

where $\mathbb{Y}_t$ and $\mathbb{X}_t$ are column vectors of order $k\times 1$. $\mathbb{Z}_t$ is a block column vector of $n-1$ vectors of length $k$ and $\mathbb{D}$ is a block row vector of length $n-1$ of square matrices of order $k\times k$. Using these variables the VECM becomes,

$
\begin{align}
\mathbb{Y}_t &= \alpha \beta^\prime \mathbb{X}_t + \mathbb{D}\mathbb{Z}_t +  \mathcal{E}_t
\end{align}
\tag{27}
$

Further define,

$
\begin{align}
\Sigma_{yx} &= \text{E}\left[\mathbb{Y}_t \mathbb{X}_t^\prime\right] \\
\Sigma_{xy} &= \text{E}\left[\mathbb{X}_t \mathbb{Y}_t^\prime\right] = \Sigma_{yx}^\prime \\
\Sigma_{zy} &= \text{E}\left[\mathbb{Z}_t \mathbb{Y}_t^\prime\right] = \Sigma_{yz}^\prime \\
\Sigma_{xz} &= \text{E}\left[\mathbb{X}_t \mathbb{Z}_t^\prime\right] = \Sigma_{zx}^\prime \\
\end{align}
\tag{28}
$

$
\begin{align}
\Sigma_{xx} &= \text{E}\left[\mathbb{Y}_t \mathbb{X}_t^\prime\right] \\
\Sigma_{yy} &= \text{E}\left[\mathbb{X}_t \mathbb{Y}_t^\prime\right] \\
\Sigma_{zz} &= \text{E}\left[\mathbb{Z}_t \mathbb{Y}_t^\prime\right]
\end{align}
\tag{29}
$

Define the transformations,

$
\begin{align}
\mathbb{X}_t^* &= \mathbb{X}_t - \Sigma_{xz}\Sigma^{-1}_{zz}\mathbb{Z}_t \\
\mathbb{D}^* &= \mathbb{D} + \alpha\beta^\prime\Sigma_{xz}\Sigma^{-1}_{zz}
\end{align}
\tag{30}
$

It is seen that $\mathbb{X}_t^*$ is the residual of the OLS regression of $\mathbb{X}_t$ on $\mathbb{Z}_t$ where $\Sigma_{xz}\Sigma^{-1}$ is the regression coefficient. Note that $\mathbb{X}_t^*$  and $\mathbb{Z}_t$ are uncorrelated,

$
\begin{align}
\text{E}\left[\mathbb{X}_t^* \mathbb{Z}^\prime_t\right] = \Sigma_{xz} - \Sigma_{xz}\Sigma_{zz}^{-1} \Sigma_{zz} = 0
\end{align}
\tag{31}
$

It follows that,

$
\begin{align}
\mathbb{Y}_t &= \alpha \beta^\prime \mathbb{X}^*_t + \mathbb{D}^*\mathbb{Z}_t +  \mathcal{E}_t
\end{align}
\tag{32}
$

further let,

$
\begin{align}
\mathbb{Y}^*_t &= \mathbb{Y}_t - \mathbb{D}^*\mathbb{Z}_t
\end{align}
\tag{33}
$

Regressing $\mathbb{Y}_t$ on $\mathbb{Z}_t$ using OLS gives,

$
\begin{align}
\mathbb{D}^* = \Sigma_{yz}\Sigma_{zz}^{-1}
\end{align}
\tag{31}
$

so $\mathbb{Y}^*_t$ is the residual of the regression of $\mathbb{Y}_t$ on  $\mathbb{Z}_t$. Bringing things together gives,

$
\begin{align}
\mathbb{Y}^*_t &= \alpha\beta^\prime \mathbb{X}^*_t + \mathcal{E}_t \\
\mathbb{D} &= \mathbb{D}^* - \alpha\beta^\prime\Sigma_{xz}\Sigma_{zz}^{-1} \\
&= \Sigma_{yz}\Sigma_{zz} - \alpha\beta^\prime\Sigma_{xz}\Sigma_{zz}^{-1}
\end{align}
\tag{33}
$

This is the most basic form of the reduced rank regression model. It can be shown that if $\alpha\beta^\prime$ is not of full rank both can be estimated using canonical correlation analysis. The result is,

$
\begin{align}
\alpha &= \Sigma_{yy\cdot z}^{-\frac{1}{2}} \mathbb{M^{(r)}} \\
\beta^\prime &= \mathbb{M^{(r)}}^\prime \Sigma_{yy\cdot z}^{-\frac{1}{2}}\Sigma_{yx\cdot z}\Sigma_{xx\cdot z}^{-1} 
\end{align}
\tag{34}
$

Using,

$
\begin{align}
\Sigma_{yx^*} &= \Sigma_{yx} - \Sigma_{yz}\Sigma^{-1}_{zz}\Sigma_{zx} = \Sigma_{yx\cdot z} \\
\Sigma_{xy^*} &= \Sigma_{xy} - \Sigma_{xz}\Sigma^{-1}_{zz}\Sigma_{zy} = \Sigma_{xy\cdot z} \\
\Sigma_{x^* x^*} &= \Sigma_{xx} - \Sigma_{xz}\Sigma^{-1}_{zz}\Sigma_{zx} = \Sigma_{xx\cdot z} \\
\Sigma_{y^* y^*} &= \Sigma_{yy} - \Sigma_{yz}\Sigma^{-1}_{zz}\Sigma_{zy} = \Sigma_{yy\cdot z} \\
\Sigma_{xy\cdot z} &= \Sigma_{yx\cdot z}^\prime \\
\end{align}
\tag{35}
$

where,

$
\begin{align}
\mathbb{M^{(r)}} = \left[M_1\ M_2\ \cdots\ M_r\right]
\end{align}
\tag{36}
$

and $M_i$ is the eigenvector corresponding to the $i$'th largest eigenvalue $\rho_i^2$ of the matrix,

$
\begin{align}
R = \Sigma_{yy.z}^{-\frac{1}{2}}\Sigma_{yx.z}\Sigma_{xx.z}^{-1}\Sigma_{xy.z}\Sigma_{yy.z}^{-\frac{1}{2}}
\end{align}
\tag{37}
$

additionally $\alpha$ and $\beta$ satisfy the following normalization,

$
\begin{align}
\alpha^\prime \Sigma_{yy.z} \alpha &= \mathbb{1^{(r)}} \\
\beta^\prime \Sigma_{xx.z} \beta &= \mathbb{P}^2
\end{align}
\tag{38}
$

where $\mathbb{P}^2$ is the eigenvalue matrix,

$
\begin{align}
\mathbb{P}^2 = 
\begin{bmatrix}
\rho_1^2 & 0 & \cdots & 0 \\
0 & \rho_2^2 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & \rho_r^2 \\
\end{bmatrix}
\end{align}
\tag{39}
$

The canonical variates are given by,

$
\begin{align}
U &= H \mathbb{Y}^* \\
V &= G \mathbb{X}^* \\
\end{align}
\tag{40}
$

with,

$
\begin{align}
H &= \mathbb{M^{(r)}}^\prime \Sigma_{yy.z}^{-\frac{1}{2}} \\
G &= \mathbb{M^{(r)}}^\prime \Sigma_{yy.z}^{-\frac{1}{2}}\Sigma_{yx.z}\Sigma_{xx.z}^{-1} \\
\end{align}
\tag{41}
$

## Johansen Procedure

The Johansen procedure is used to determine the rank of $\alpha\beta^\prime$ and to evaluate $\alpha$ and $\beta^\prime$ and $A_i, i=2,3,\ldots, n$ using canonical correlation analysis. If the canonical variates obtained are denoted by,

$
\begin{align}
U &= \left[U_1, U_2, \ldots, U_r\right]\\
V &=  \left[V_1, V_2, \ldots, V_r\right]\\
\end{align}
\tag{42}
$

where the correlation coefficients are the eigenvalues obtained for canonical correlation analysis and,

$
\begin{align}
\rho_i^2 &= \text{Cov}\left(U_i V_i\right) &= \text{E}\left[U_iV_i\right] \\
\end{align}
\tag{43}
$

and for $i\ne j$,

$
\begin{align}
\rho_i^2 &= \text{Cov}\left(U_i V_j\right) = 0\\
\end{align}
\tag{44}
$

The null hypothesis used by the Johansen procedure to determine rank is that $\rho_i^2 = 0$ for $i > r$ where $r$ is the rank. The test statistic is given by,

$
\begin{align}
2 \ln\lambda = T \sum_{i=r+1}^k \ln\left(1 - \rho_i^2\right)
\end{align}
\tag{45}
$

which has a $\chi^2$ distribution. The system is cointegrated if $r < k$.

## Granger Causality

Consider the two variable model,

$
\begin{align}
y_t = \alpha + \varphi y_{t-1} + \beta X_{t-1} +  \varepsilon_t
\end{align}
\tag{46}
$

If $\beta \ne 0$ then values of $x_{t-1}$ contribute to $y_t$. This is the definition of Granger Causality. It is said that $x_t$ Granger Causes $y_t$. A test for Granger Causality will test for $\beta \ne 0$. This is implemented using the OLS t-test to test for a mean value of zero and an F-test to compare the variance to the value obtained for $\beta=0$

Granger causality can determine the common trends in the model. A common trend will be Granger causal for the variables which depend on the error correction term and will not have any causal variables. The test is performed on the first stationary difference of the series.

## Time Asymptotic Properties

Time asymptotically the the error correction term dominates the VECM resulting in the equation,

$
\begin{align}
\beta^\prime X_t = 0
\end{align}
\tag{47}
$

This gives a system of linear equations which can be solved for $\beta$. For example for a three variable model with two cointegrating vectors the following system of equations is obtained,

$
\begin{align}
\begin{bmatrix}
1 & -\beta_{12} & -\beta_{13} \\
-\beta_{21} & 1 & -\beta_{23}
\end{bmatrix}
\begin{bmatrix}
x_{1,t} \\
x_{2,t} \\
x_{2,t}
\end{bmatrix}
&= 
\begin{bmatrix}
0 \\
0
\end{bmatrix}
\end{align}
\tag{48}
$

and for a single cointegrating vector the following equation is obtained,

$
\begin{align}
x_{1,t} - \beta_2 x_{2,t} - \beta_3 x_{3,t} = 0
\end{align}
\tag{49}
$

## Analysis Procedure

The following modules are used in the analysis.

```
from statsmodels.tsa.vector_ar import vecm
from statsmodels.tsa.stattools import grangercausalitytests
from statsmodels.tsa.vector_ar.var_model import VAR
from statsmodels.tsa.vector_ar.vecm import VECM
```

The steps to perform the analysis are as follows,

1. ADF stationarity test all series and difference series until the test pass. This number of differences will define the integration 
order of the system. VECM models are $I(1)$ so should be stationary after one difference.
2. Estimate model lags using `VAR.select_order`. Verify with ACF-PACF. 
3. After the first two steps it will be known that the series are $I(1)$ and the number of lags will be known. `grangercausalitytests` can be applied to the first differences to determine the common trends and suggest the model rank.
4. Use the Johansen test from `vecm.coint_johansen` to determine the rank of the model, This should be consistent with the Granger causality results.
5. Now estimates for the rank and lag are known and the model parameters can be estimated using `VECM.fit()`. OLS can be used to estimate $\beta$ vectors using time asymptotic form.
6. Verify that residual is stationary using ADF test.

````python
def adf_test(xt):
    for i in range(len(xt)):
        result, _ = adf.compute_adf_test(xt[i])
        if result.status_vals[1]:
            print("Not Stationary")
        else:
            print("Stationary")

def plot_vecm_simulation(t, xt, title):
    labels = [r"$x_1$", r"$x_2$", r"$x_3$"]   
    comparison(xt, t, labels =labels, title=title, xlabel=r"$t$", lw=1, title_offset=0.05)

def plot_vecm_dff(dxt, title):
    labels = [r"$\Delta x_1$", r"$\Delta x_2$", r"$\Delta x_3$"]   
    stack(dxt, title=title, ylabels=labels, xlabel=r"$t$", lw=1, title_offset=0.05, figsize=(10, 8))

def pacf_acf_plot(t, dxt, nlags, title):
    acf = [stats.compute_acf(t, samp, nlags=nlags) for samp in dxt] 
    pacf = [arima.compute_pacf(samp, nlags=nlags) for samp in dxt]
    data = [[acf[i][1], pacf[i][1]] for i in range(len(dxt))]
    time = [[acf[i][0], pacf[i][0]] for i in range(len(dxt))]
    comparison_stack(data, time, title=title, xlabel=r"Lag", ylim=(-1.0, 1.0), figsize=(10, 8), 
                     labels=[["ACF", "PACF"]], ylabels=[r"$\Delta x_1$", r"$\Delta x_2$", r"$\Delta x_3$"])
````
